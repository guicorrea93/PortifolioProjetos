```python
# ═══════════════════════════════════════════════════════════════════════════
# CÉLULA 4: Criar estrutura de pastas
# ═══════════════════════════════════════════════════════════════════════════

print("\n" + "="*80)
print("📁 CRIANDO ESTRUTURA DE PASTAS")
print("="*80)
print()

from pathlib import Path

# Definir estrutura
folders = [
    'data/raw',
    'data/processed',
    'outputs/figures',
    'outputs/results',
    'outputs/tables',
    'presentation'
]

# Criar pastas
for folder in folders:
    Path(folder).mkdir(parents=True, exist_ok=True)
    print(f"✅ Criada: {folder}/")

print("\n✅ Estrutura de pastas criada!")

# Verificar estrutura
print("\n📂 Estrutura do projeto:")
for folder in sorted(folders):
    level = folder.count('/')
    indent = "  " * level
    folder_name = folder.split('/')[-1]
    print(f"{indent}└── 📁 {folder_name}")
```

```python
# ═══════════════════════════════════════════════════════════════════════════
# CÉLULA 5: Configurar visualizações
# ═══════════════════════════════════════════════════════════════════════════

print("\n" + "="*80)
print("🎨 CONFIGURANDO VISUALIZAÇÕES")
print("="*80)
print()

# Configurações do Matplotlib
plt.style.use('seaborn-v0_8-darkgrid')
plt.rcParams['figure.figsize'] = (12, 6)
plt.rcParams['font.size'] = 10
plt.rcParams['axes.labelsize'] = 12
plt.rcParams['axes.titlesize'] = 14
plt.rcParams['xtick.labelsize'] = 10
plt.rcParams['ytick.labelsize'] = 10
plt.rcParams['legend.fontsize'] = 10

print("✅ Matplotlib configurado")
print(f"   - Estilo: seaborn-darkgrid")
print(f"   - Tamanho padrão: 12x6 inches")

# Configurações do Seaborn
sns.set_palette("husl")
sns.set_context("notebook")

print("✅ Seaborn configurado")
print(f"   - Paleta: husl")
print(f"   - Contexto: notebook")

# Configurações do Plotly
import plotly.io as pio
pio.templates.default = "plotly_white"

print("✅ Plotly configurado")
print(f"   - Template: plotly_white")

# Configurações do Pandas
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.float_format', '{:.2f}'.format)
pd.set_option('display.width', 200)

print("✅ Pandas configurado")
print(f"   - Exibir todas as colunas")
print(f"   - Formato de float: 2 casas decimais")

# Seed para reprodutibilidade
RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

print(f"\n✅ Seed configurado: {RANDOM_STATE}")
print("   (garante resultados reproduzíveis)")
```

```python
# ═══════════════════════════════════════════════════════════════════════════
# CÉLULA 6: Testar criação de gráfico
# ═══════════════════════════════════════════════════════════════════════════

print("\n" + "="*80)
print("🧪 TESTE DE VISUALIZAÇÃO")
print("="*80)
print()

# Criar dados de teste
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Teste Matplotlib
fig, axes = plt.subplots(1, 2, figsize=(14, 4))

axes[0].plot(x, y, linewidth=2, color='#FF6B6B')
axes[0].set_title('Teste Matplotlib', fontsize=14, fontweight='bold')
axes[0].set_xlabel('X')
axes[0].set_ylabel('sin(X)')
axes[0].grid(True, alpha=0.3)

# Teste Seaborn
sns.lineplot(x=x, y=y, ax=axes[1], linewidth=2, color='#4ECDC4')
axes[1].set_title('Teste Seaborn', fontsize=14, fontweight='bold')
axes[1].set_xlabel('X')
axes[1].set_ylabel('sin(X)')

plt.tight_layout()
plt.show()

print("✅ Gráficos Matplotlib e Seaborn funcionando!")

# Teste Plotly
import plotly.graph_objects as go

fig = go.Figure()
fig.add_trace(go.Scatter(
    x=x, 
    y=y, 
    mode='lines',
    name='sin(x)',
    line=dict(color='#95E1D3', width=3)
))

fig.update_layout(
    title='Teste Plotly (Interativo)',
    xaxis_title='X',
    yaxis_title='sin(X)',
    height=400,
    template='plotly_white'
)
fig.show()

print("✅ Gráficos Plotly funcionando!")
```
